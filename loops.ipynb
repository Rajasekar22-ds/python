```python
def vowels(words):
    count=0
    for i in (words):
        if i in "aeiou":
            print(i)
            count+=1
    return count
       

words=input("enter a word")
count=vowels(words)
print ("the total number of vowels present in your sentence({}) is {}".format(words,count)) 
```

## USER DEFINED FUNCTION WITH FOR AND IF (break)

```python
def login(username,passward):
    attept=5
    for i in range(1,attept):
        username1=input("enter user name :")
        passward1=input("enter passward")
        if username==username1 and passward==passward1:
            print("dear user {} \n your login successful".format(username))
            break
        else :
            attept-=1
            print("invalid username or passward")
            print("you have",attept,"only attempts") 


username="raja"
passward="raja@123"
login(username,passward)
       
```

```python
def findnum(number, list1):
    if number in list1:
        for i in range(len(list1)):
            if list1[i] == number:
                print("Number found in list")
                print("Position:", i)
                break
    else:
        print("Number not found in list")

list1 = [83, 14, 42, 67, 5, 91, 28, 56, 73, 39,
         94, 22, 49, 61, 8, 88, 31, 70, 52, 17]

number = int(input("Enter a number: "))

findnum(number, list1)
```
