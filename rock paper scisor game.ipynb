## Rock Paper Scissors

```python
import random
options=["rock","paper","scissor"]
botoption=random.choice(options)
exit="a"
def game(options,botoption,exit):
    while exit!="x":
            userinput=input('eneter your option ("rock" or "paper" or ""scissor)').lower()
            try:
                print("you :",userinput)
                print("computer :",botoption)
                if userinput not in options:
                    raise ValueError("Invalid choice")
                elif botoption==userinput:
                    print ("match draw")
                elif botoption=="rock" and userinput=="paper":
                    print("you win")
                elif botoption=="rock" and userinput=="scissor":
                    print("you loose")
                elif botoption=="paper" and userinput=="scissor":
                    print("you win")
                elif botoption=="paper" and userinput=="rock":
                    print("you loose")
                elif botoption=="scissor" and userinput=="rock":
                    print("you win")
                elif botoption=="scissor" and userinput=="paper":
                    print("you loose")
            except ValueError:
                print("eneter a valid options")
            optionexit=input("""    if you want to exit press 'x'
                to continue press 'eneter'""")
            exit=optionexit
game(options,botoption,exit)
```
