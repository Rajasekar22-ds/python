# 🧮 Math Quiz Game

This is a simple Math Quiz Game created using Python. The game asks random math questions and checks whether the answer is correct. At the end, it shows the final score and percentage. This project helped me practice basic Python concepts in a fun and interactive way.

## Python Concepts Used

- Variables
- User Input
- Functions
- For Loop
- If-Else Statements
- Random Module
- Exception Handling
- Score Calculation

##  How the Game Works

1. The game creates random math questions.
2. The player enters an answer for each question.
3. The program checks whether the answer is correct.
4. One point is awarded for every correct answer.
5. After all questions are completed, the final score and percentage are displayed.
6. The player can choose to play another round.

```python
import random
```

```python
def generate_question():
    
    operations = ["+", "-", "*"]

    num1 = random.randint(1,20)
    num2 = random.randint(1,20)

    operation = random.choice(operations)

    if operation == "+":
        answer = num1 + num2

    elif operation == "-":
        answer = num1 - num2

    else:
        answer = num1 * num2

    return num1, num2, operation, answer
```

```python
def play_quiz():

    print("="*50)
    print("🧮 Welcome to Math Quiz Game")
    print("="*50)

    score = 0
    total_questions = 10

    for question in range(1,total_questions+1):

        num1,num2,operation,answer = generate_question()

        print(f"\nQuestion {question}")

        print(f"{num1} {operation} {num2} = ?")

        try:

            user_answer = int(input("Enter your answer : "))

            if user_answer == answer:
                print("✅ Correct Answer")
                score += 1

            else:
                print(f"❌ Wrong Answer")
                print(f"Correct Answer : {answer}")

        except ValueError:

            print("Invalid Input")
            print(f"Correct Answer : {answer}")

    percentage = (score/total_questions)*100

    print("\n"+"="*50)
    print("Quiz Completed")
    print("="*50)

    print(f"Total Score : {score}/{total_questions}")
    print(f"Percentage  : {percentage:.2f}%")

    if percentage == 100:
        print("🏆 Excellent!")

    elif percentage >= 80:
        print("🌟 Very Good!")

    elif percentage >= 60:
        print("👍 Good Job!")

    else:
        print("📚 Keep Practicing!")
```

```python
while True:

    play_quiz()

    choice = input("\nDo you want to play again? (yes/no): ").lower()

    if choice != "yes":
        print("\n👋 Thank You for Playing!")
        break
```
