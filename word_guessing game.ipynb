## Word Guessing Game

```python

import random
print("Welcome to Word Guessing Game\n")

def wordguessing():
    words = [
        "cat", "dog", "sun", "hat", "pen",
    "book", "fish", "ball", "tree", "milk",
    "apple", "mango", "grape", "house", "chair",
    "table", "water", "bread", "clock", "phone"
    ]
    level=1
    i=0
    while True:
        random_word=random.choice(words)
        attempt=3
        result=""
        z=0
        for i in range (len(random_word)):
            if i%2==0:
                result+=random_word[i]
            else:
                result+=" _ "
        print("level ",level)
        print ("Find :",result)
        try:
            while z<=3:
                userinput=input("enter word")
                if userinput.isalpha()==False:
                    raise ValueError
                if userinput.lower()==random_word.lower():
                    print("congradulation!!\n You have guessed the correct answer: x ",random_word)
                    level+=1
                    break
                else:
                    print("wrong answer!! chance left :",attempt)
                    attempt-=1
                    z+=1
            exitcontrol=input(
                "Press Enter to continue or X to exit: "
            ).upper()  
            if exitcontrol=="X":
                print("thanks for playing")
                break
        except ValueError:
            print("Enter alphabet only!!\n No numbers or sybools allowed")
wordguessing()
```
